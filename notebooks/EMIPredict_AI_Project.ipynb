```python
# Temporary numeric conversion
age_num = pd.to_numeric(df["age"], errors="coerce")
salary_num = pd.to_numeric(df["monthly_salary"], errors="coerce")
bank_num = pd.to_numeric(df["bank_balance"], errors="coerce")

checks = {
    "age": age_num,
    "monthly_salary": salary_num,
    "credit_score": df["credit_score"],
    "bank_balance": bank_num,
    "max_monthly_emi": df["max_monthly_emi"]
}

for name, series in checks.items():
    
    print(name.upper())
    
    print(f"Minimum: {series.min()}")
    print(f"Maximum: {series.max()}")
    print(f"Missing after numeric conversion: {series.isna().sum():,}")
```

```python
numeric_columns = [
    "age",
    "monthly_salary",
    "bank_balance"
]

for col in numeric_columns:

    # Convert to string for cleaning
    df[col] = df[col].astype(str)

    # Remove extra repeated ".0"
    # Example: "18000.0.0" → "18000.0"
    df[col] = df[col].str.replace(
        r"\.0\.0$",
        ".0",
        regex=True
    )

    # Convert invalid strings to NaN
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

print("Numeric columns cleaned successfully!")

print("\nUpdated Data Types:")
print(df[numeric_columns].dtypes)

print("\nMissing Values After Cleaning:")
print(df[numeric_columns].isnull().sum())
```

```python
missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (
        df.isnull().sum() / len(df) * 100
    ).round(3)
})

missing_summary = missing_summary[
    missing_summary["Missing Values"] > 0
].sort_values("Missing Values", ascending=False)

display(missing_summary)
```

```python
# 1. Education - fill with most frequent value
education_mode = df["education"].mode()[0]

df["education"] = df["education"].fillna(
    education_mode
)


# 2. Monthly salary - fill with median salary
# based on employment type
df["monthly_salary"] = df["monthly_salary"].fillna(
    df.groupby("employment_type")["monthly_salary"]
      .transform("median")
)


# 3. Monthly rent - fill with median rent
# based on house type
df["monthly_rent"] = df["monthly_rent"].fillna(
    df.groupby("house_type")["monthly_rent"]
      .transform("median")
)


# 4. Credit score - fill with median
df["credit_score"] = df["credit_score"].fillna(
    df["credit_score"].median()
)


# 5. Bank balance - fill with median
df["bank_balance"] = df["bank_balance"].fillna(
    df["bank_balance"].median()
)


# 6. Emergency fund - fill with median
df["emergency_fund"] = df["emergency_fund"].fillna(
    df["emergency_fund"].median()
)


print("Missing values handled successfully!")
```

```python
missing_values = df.isnull().sum()

remaining_missing = missing_values[
    missing_values > 0
]

print("Total missing values:", missing_values.sum())

if remaining_missing.empty:
    print("✅ No missing values remaining!")
else:
    print("\nRemaining missing values:")
    print(remaining_missing)
```

```python
range_checks = {
    "age": (df["age"], 18, 100),
    "monthly_salary": (df["monthly_salary"], 0, 500000),
    "years_of_employment": (df["years_of_employment"], 0, 50),
    "monthly_rent": (df["monthly_rent"], 0, 100000),
    "credit_score": (df["credit_score"], 300, 850),
    "bank_balance": (df["bank_balance"], 0, 2000000),
    "emergency_fund": (df["emergency_fund"], 0, 1000000),
    "requested_amount": (df["requested_amount"], 0, 2000000),
    "requested_tenure": (df["requested_tenure"], 1, 120),
    "max_monthly_emi": (df["max_monthly_emi"], 0, 100000)
}

results = []

for column, (series, min_val, max_val) in range_checks.items():

    below_min = (series < min_val).sum()
    above_max = (series > max_val).sum()

    results.append({
        "Column": column,
        "Min Value": series.min(),
        "Max Value": series.max(),
        "Below Minimum": below_min,
        "Above Maximum": above_max
    })

range_results = pd.DataFrame(results)

display(range_results)
```

```python
invalid_credit = df[
    (df["credit_score"] < 300) |
    (df["credit_score"] > 850)
]

print("Total invalid credit score records:", len(invalid_credit))

print("\nInvalid credit score value distribution:")
print(
    invalid_credit["credit_score"]
    .value_counts()
    .sort_index()
    .head(30)
)

print("\nLowest invalid values:")
print(
    df.loc[df["credit_score"] < 300, "credit_score"]
    .value_counts()
    .sort_index()
    .head(20)
)

print("\nHighest invalid values:")
print(
    df.loc[df["credit_score"] > 850, "credit_score"]
    .value_counts()
    .sort_index()
    .tail(20)
)
```

```python
invalid_credit_mask = (
    (df["credit_score"] < 300) |
    (df["credit_score"] > 850)
)

print("Invalid credit scores before cleaning:",
      invalid_credit_mask.sum())

df.loc[invalid_credit_mask, "credit_score"] = np.nan


# Calculate median using valid credit scores only
valid_credit_median = df["credit_score"].median()

print("Median of valid credit scores:",
      valid_credit_median)


# Fill invalid/missing values
df["credit_score"] = df["credit_score"].fillna(
    valid_credit_median
)

print("\nCredit score cleaned successfully!")
```

```python

print("Missing credit scores:",
      df["credit_score"].isnull().sum())

print("Minimum credit score:",
      df["credit_score"].min())

print("Maximum credit score:",
      df["credit_score"].max())

print("\nInvalid below 300:",
      (df["credit_score"] < 300).sum())

print("Invalid above 850:",
      (df["credit_score"] > 850).sum())
```

```python
df["gender"] = (
    df["gender"]
    .astype(str)
    .str.strip()
    .str.upper()
)

# Map all variations to standard values
df["gender"] = df["gender"].replace({
    "M": "Male",
    "MALE": "Male",
    "F": "Female",
    "FEMALE": "Female"
})

print("Gender values after standardization:")
print(df["gender"].value_counts())
```
